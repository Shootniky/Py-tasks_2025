1. Очистить данные в представленном датасете, удалить строчки с пустыми значениями, привести данные столбца PuFrom к одному виду (имеются записи с разным регистром).


```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("Uber.csv")
```

```python
df["PuFrom"] = df["PuFrom"].fillna("nan")
```

```python
df["Address"] = df["Address"].replace('      ', np.nan)
```

```python
df["Address"] = df["Address"].fillna("nan")
```

```python
df_clean = df.drop(index=df[df["PuFrom"] == "nan"].index)
```

```python
df_clean = df.drop(index=df[df["Address"] == "nan"].index)
```

```python
df_clean["PuFrom"] = df_clean["PuFrom"].astype(str)
```

```python
df_clean["PuFrom"]= df_clean["PuFrom"].str.upper()
```



2. Перевести значение столбца Time (время поездки) во время в минутах (в таблице указаны минуты:секунды)
То есть 14:30 нужно привести к виду 14.5, а 20:15 к виду 20.25

```python
def convertion(time):
  minutes, seconds = map(int, time.split(':'))
  total_minutes = minutes + seconds / 60
  return total_minutes
```

```python
df_clean["Time"] = df_clean["Time"].apply(convertion)
```
