Написать программу, которая сможет:
1. Подключиться к сайту
2. Получить данные об автомобилях из области спецразмещения (получить марку автомобиля и стоимость). 
3. Поместить данные в DataFrame 
4. Область спецразмещения находится в верхней части сайта и выделена жёлтым цветом.
Обратите внимание на то, что цена и модель автомобиля находятся в одном контейнере, но в разных тегах.

https://auto.drom.ru/

```python
import pandas as pd
import requests as rg
import bs4
```

```python
def drom_info(url: str) -> pd.DataFrame:
    response = rg.get(url)
    
    if response.status_code != 200:
        print(f"Error while loading the page: {response.status_code}")
        return pd.DataFrame()

    soup = bs4.BeautifulSoup(response.content, 'html.parser')
    
    car_models = soup.find_all('div', class_='u41pni2')  # Model
    car_prices = soup.find_all('div', class_='u41pni8')  # Price

    models_list = [model.get_text(strip=True) for model in car_models]
    prices_list = [price.get_text(strip=True) for price in car_prices]
    
    # Создаем DataFrame
    df_cars = pd.DataFrame({
        'Model': models_list,
        'Price in rubles': prices_list
    })

    return df_cars

drom_url = 'https://auto.drom.ru/'
car_data = drom_info(drom_url)
```

```python
car_data["Price in rubles"] = car_data["Price in rubles"].str.replace("₽", "")
car_data["Price in rubles"] = car_data["Price in rubles"].str.replace("\xa0", "", regex=False)
```

```python
car_data["Price in rubles"] = car_data["Price in rubles"].astype(float)
```
